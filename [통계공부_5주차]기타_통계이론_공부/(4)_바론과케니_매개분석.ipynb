# <바론과 케니 매개분석>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import koreanize_matplotlib
import os
import warnings

warnings.filterwarnings("ignore")
```

```python
# Seed for reproducibility
np.random.seed(42)

# 데이터 생성
n = 200
X = np.random.normal(5, 2, n)          # 운동 시간
M = 0.5 * X + np.random.normal(0, 1, n) # 운동시간이 체력수준에 영향을 줌
Y = 0.6 * M + 0.3 * X + np.random.normal(0, 1, n) # 체력수준과 운동시간 둘 다 건강에 영향

df = pd.DataFrame({'exercise_time': X, 'fitness': M, 'health': Y})

df.head()
```

```python
### 원본 데이터
df
```

# 1. 바론과 케니의 4단계 검증법을 적용하여 매개효과를 검증하시오.

- 1단계: exercise_time → health
- 2단계: exercise_time → fitness
- 3단계: fitness → health (exercise_time 통제)
- 4단계: 매개효과 확인

```python
### 1단계 : exercise_time → health
from statsmodels.api import OLS

formula = "health ~ exercise_time"

model1 = OLS.from_formula(formula, data=df).fit()
model1.summary()
```

```python
### 2단계 : exercise_time → fitness
from statsmodels.api import OLS

formula = "fitness ~ exercise_time"

model2 = OLS.from_formula(formula, data=df).fit()
model2.summary()
```

```python
### 3단계 : fitness → health (exercise_time 통제)
from statsmodels.api import OLS

formula = "health ~ exercise_time + fitness"

model3 = OLS.from_formula(formula, data=df).fit()
model3.summary()
```

```python
### 4단계 (매개효과 확인)

### 총효과 (Total Effect)
te = model1.params["exercise_time"]     ## model1의 독립변수의 회귀계수

### 직접효과 (Direct Effect)
de = model3.params["exercise_time"]    ## model3의 독립변수의 회귀계수


te > de
```

```python
print("매개효과 존재한다.")
```

# 2. 다음을 계산하시오:

- 총효과 크기

```python
### 총효과 (Total Effect)

te = model1.params["exercise_time"]     ## model1의 독립변수의 회귀계수
te
```

```python
### 직접효과 (Direct Effect)

de = model3.params["exercise_time"]    ## model3의 독립변수의 회귀계수
de
```

```python
### 간접효과(=매개효과) (Indirect Effect)

a = model2.params["exercise_time"]     ## model2의 독립변수의 회귀계수
b = model3.params["fitness"]           ## model3의 매개변수의 회귀계수

ie = a*b
ie
```

```python
de +ie
```

# 3. 매개효과의 유의성을 검정하고, 완전매개인지 부분매개인지 판단하시오.

```python
### Sobel Test(간접효과 통계적 유의성 검정)
from scipy.stats import norm

# H0 : 매개효과가 존재하지 않는다.
# H1 : 매개효과가 존재한다.

# 간접효과 (Indirect Effect)
a = model2.params["exercise_time"]     ## model2의 독립변수의 회귀계수
b = model3.params["fitness"]           ## model3의 매개변수의 회귀계수

ie = a*b

# 표준오차(Standard Error)
se_a = model2.bse['exercise_time']
se_b = model3.bse['fitness']

# Sobel 검정
se_indirect = np.sqrt( (a**2 * se_b**2) + (b**2 * se_a**2) )
z_stat = ie / se_indirect

rv = norm(loc=0, scale=1)
p_value = 2 * (1 - rv.cdf(abs(z_stat)))

print(f"검정통계량 : {z_stat}")
print(f"검정통계량의 p-value : {p_value}")
print("귀무가설 채택" if p_value > 0.05 else "귀무가설 기각")
```

```python
print("매개효과가 존재한다.")
```

```python
### 완전매개, 부분매개 구분
model3.pvalues["exercise_time"] < 0.05
```
