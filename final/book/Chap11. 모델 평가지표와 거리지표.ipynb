# 11-1. 회귀모델 평가지표


## 1. ME (Mean Error, 평균 오차)

- **공식:** `ME = (1/n) × Σ(yi - ŷi)`
- **설명:** 예측값과 실제값 차이의 평균, 모델 편향(bias) 측정
- **범위/해석:** 음수~양수, 0에 가까울수록 좋음 (양수=과소예측, 음수=과대예측)
- **장점:** 
  - 계산 간단
  - 모델 편향성 직관적 파악 가능
- **단점:** 
  - 양의 오차와 음의 오차 상쇄로 절대 오차 크기 파악 어려움
- **사용예:** 모델의 전반적인 편향성(과소/과대 예측 여부) 확인

## 2. MAE (Mean Absolute Error, 평균 절대 오차)

- **공식:** `MAE = (1/n) × Σ|yi - ŷi|`
- **설명:** 에러에 절대값을 취하여 에러 크기를 그대로 반영
- **범위/해석:** 0 이상, 0에 가까울수록 좋음
- **장점:** 
  - 직관적 해석 가능
  - 이상치 영향 최소화
- **단점:** 
  - 잔차 절대값으로 과소/과대예측 파악 어려움
  - 스케일 의존적
  - 미분 불연속점 존재로 최적화 어려움
- **사용예:** 이상치에 덜 민감한 평가, 선형 손실 함수

## 3. MAPE (Mean Absolute Percentage Error, 평균 절대 백분율 오차)

- **공식:** `MAPE = (100/n) × Σ|(yi - ŷi)/yi|`
- **설명:** MAE를 퍼센트로 변환, 예측값 대비 잔차 비율, 주로 시계열 모형 비교에 사용
- **범위/해석:** 0~∞%, 작을수록 적합도 높음
- **장점:** 
  - 이해하기 쉬움
  - 비율 변수로 모델간 성능 비교 용이
  - 단위 무관 비교 가능
- **단점:** 
  - 실제값 대비 과소/과대예측 여부 파악 어려움
  - 비율 해석 시 의미있는 값에만 적용 가능
  - 실제값 0 포함 시 사용 불가
  - 실제값이 0에 가까운 경우 MAPE 값 급증
- **사용예:** 서로 다른 스케일 데이터 비교, 상대적 오차 평가

## 4. MPE (Mean Percentage Error, 평균 백분율 오차)

- **공식:** `MPE = (100/n) × Σ((yi - ŷi)/yi)`
- **설명:** MAPE 절대값 제외 지표, 표준화로 스케일링 문제 해결, 0 근접 시 모델 적합 우수
- **범위/해석:** 음수~양수, 0에 가까울수록 좋음
- **장점:** 
  - 과소/과대 예측 방향 파악 가능 (양수=과소예측, 음수=과대예측)
  - 편향 측정 가능
- **단점:** 
  - 지표 자체 직관성 부족
  - 예측변수와 단위 상이
  - 절대 오차 크기 정보 상실
- **사용예:** 과소/과대 예측 편향 확인, 상대적 편향 측정

## 5. MSE (Mean Squared Error, 평균 제곱 오차)

- **공식:** `MSE = (1/n) × Σ(yi - ŷi)²`
- **설명:** 예측값과 실제값 제곱 차이의 평균, 지표 자체 직관적이나 예측변수와 단위 상이
- **범위/해석:** 0 이상, 0에 가까울수록 좋음
- **장점:** 
  - 미분 연속으로 최적화 용이
- **단점:** 
  - 이상치에 민감
  - 스케일 의존적
  - 과소/과대 예측 여부 파악 불가
- **사용예:** 이상치 민감 평가, 큰 오차 패널티 부여

## 6. RMSE (Root Mean Squared Error, 평균 제곱근 오차)

- **공식:** `RMSE = √[(1/n) × Σ(yi - ŷi)²]`
- **설명:** MSE의 제곱근, 최소화 시 최대가능도 방법 결과와 동일, 원래 단위 해석 가능
- **범위/해석:** 0 이상, 0에 가까울수록 좋음
- **장점:** 
  - 직관적 비교 가능
  - 원래 데이터와 단위 일치
- **단점:** 
  - 과소/과대예측 여부 파악 어려움
  - 스케일 의존적
  - 이상치에 민감
- **사용예:** 가장 일반적인 회귀 평가, 직관적 해석 필요 시

## 7. R² (R-squared, 결정계수)

- **공식:** `R² = 1 - (SSE/SST) = 1 - (Σ(yi-ŷi)²/Σ(yi-ȳ)²)`
- **설명:** 모델이 설명하는 분산 비율
- **범위/해석:** 0~1, 1에 가까울수록 좋음
- **장점:** 
  - 표준화된 비교 지표
  - 모델 설명력 직관적 파악
- **단점:** 
  - 비선형 데이터 부적합
- **사용예:** 모델 설명력 평가, 상대적 성능 비교

## 8. MSLE (Mean Squared Logarithmic Error, 평균 제곱 로그 오차)

- **공식:** `MSLE = (1/n) × Σ[log(yi+1) - log(ŷi+1)]²`
- **설명:** MSE에 로그 적용, 로그 변환된 값들의 제곱 오차 평균
- **범위/해석:** 0 이상, 0에 가까울수록 좋음
- **장점:** 
  - 이상치 영향 감소
  - 상대적 오차 중시
- **단점:** 
  - 로그 변환으로 음수값 적용 불가
- **사용예:** 양수 데이터 상대적 오차 중시, 이상치 완화

## 9. RMSLE (Root Mean Squared Logarithmic Error, 평균 제곱근 로그 오차)

- **공식:** `RMSLE = √[(1/n) × Σ[log(yi+1) - log(ŷi+1)]²]`
- **설명:** 실제값과 예측값 로그 취한 후 차의 제곱평균제곱근 계산, RMSE 최소화 시 RMSLE도 최소화
- **범위/해석:** 0 이상, 0에 가까울수록 좋음
- **장점:** 
  - 과소예측에 큰 패널티
  - 상대적 오차 측정
  - 이상치에 강건
- **단점:** 
  - 과대예측 패널티 상대적으로 약함
- **사용예:** 과소평가 큰 패널티 필요 시(금융, 판매량 예측)

## 10. AIC (Akaike Information Criterion, 아카이케 정보기준)

- **공식:** `AIC = 2k - 2ln(L)`
- **설명:** 선형회귀모형 성능 비교, 모형과 데이터 확률분포간 Kullback-Leibler 수준 최대화 시도
- **범위/해석:** 낮을수록 좋음
- **장점:** 
  - 과적합 억제
  - 모델간 비교 용이
- **단점:** 
  - 샘플 크기 무시
  - 작은 데이터셋 과소평가 가능
- **사용예:** 모델 선택, 변수 선택, 과적합 방지

## 11. BIC (Bayesian Information Criterion, 베이지안 정보기준)

- **공식:** `BIC = ln(n)k - 2ln(L)`
- **설명:** 변수 추가에 더 강한 벌점, AIC보다 모델 복잡도에 큰 패널티
- **범위/해석:** 낮을수록 좋음
- **장점:** 
  - 보수적 과적합 방지
  - 샘플 크기 클수록 적절한 패널티
- **단점:** 
  - 작은 데이터셋 과도한 벌점 가능
- **사용예:** 대용량 데이터 모델 선택, 보수적 모델 선택

**참고사항:**
- **k**: 모델 파라미터 수
- **n**: 데이터 포인트 수  
- **L**: 최대 우도(likelihood)
- **AIC vs BIC**: BIC는 표본 크기 클수록 복잡한 모형 강하게 처벌, 큰 데이터셋→BIC 선호, 작은 데이터셋→AIC 선호

```python
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.datasets import load_diabetes
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
import koreanize_matplotlib

warnings.filterwarnings('ignore')

#load data
x, y = load_diabetes()['data'], load_diabetes()['target']

#전처리
sc = StandardScaler()
train_X, test_X, train_Y, test_Y = train_test_split(x, y)

train_X = sc.fit_transform(train_X)
test_X = sc.transform(test_X)

#선형회귀
lr = LinearRegression()
lr.fit(train_X, train_Y)
pred1 = lr.predict(test_X)

#svm 
svr = SVR()
svr.fit(train_X, train_Y)
pred2= svr.predict(test_X)

#모델평가
from sklearn.metrics import mean_absolute_error, mean_squared_error, mean_squared_log_error, r2_score

title = ['MAE', 'MAPE', 'MSE', 'RMSE', 'R2', 'MSLE,', 'RMSLE']
model_list = [('linear', pred1), ('svm', pred2)]

rs = pd.DataFrame(columns=['모델명', '성능지표명', '값'])

for m_name, p in model_list:
    mae = mean_absolute_error(test_Y, p)
    mape = (np.abs((test_Y - p)/test_Y)).mean()*100
    mse = mean_squared_error(test_Y, p)
    rmse = np.sqrt(mae)
    r2 = r2_score(test_Y, p)
    msle = mean_squared_log_error(test_Y, p)
    rmsle = np.sqrt(mean_squared_log_error(test_Y, p))
    
    r_list = [mae, mape, mse, rmse, r2, msle, rmsle]
    
    for t, r in zip(title, r_list):
        rs.loc[len(rs)] = [m_name, t, r]

g1 = rs[rs['모델명'] == 'linear'].reset_index(drop=True)
g2 = rs[rs['모델명'] == 'svm'].reset_index(drop=True)

display(pd.concat([g1, g2], axis=1))

fig, ax = plt.subplots(1, 1, figsize=(8, 6))
sns.lineplot(data=g1, x='성능지표명', y='값', marker='o', ax=ax, label='linear')
sns.lineplot(data=g2, x='성능지표명', y='값', marker='D', ax=ax, label='svm')
plt.show()
```

linear 모델의 성능지표가 더 뛰어나다. 

# 11-2. 분류모델 평가지표

## 1. Accuracy (정확도)

- **공식:** `Accuracy = (TP + TN) / (TP + TN + FP + FN)`
- **설명:** 전체 예측 중 올바르게 분류된 비율
- **범위/해석:** 0~1, 1에 가까울수록 좋음
- **장점:** 
  - 계산 간단하고 직관적 해석
  - 가장 기본적인 성능 지표
- **단점:** 
  - 불균형 데이터에서 오해 소지
  - 클래스별 성능 차이 파악 어려움
- **사용예:** 균형잡힌 데이터셋의 전반적 성능 평가

## 2. Precision (정밀도)

- **공식:** `Precision = TP / (TP + FP)`
- **설명:** 모델이 Positive로 예측한 것 중 실제 Positive인 비율
- **범위/해석:** 0~1, 1에 가까울수록 좋음
- **장점:** 
  - False Positive 최소화 중요할 때 유용
  - Positive 예측의 정확성 측정
- **단점:** 
  - 과소예측 편향 가능성
  - Recall과 상충 관계
- **사용예:** 스팸 메일 분류, 의료 진단 확진

## 3. Recall (재현율) = Sensitivity (민감도)

- **공식:** `Recall = TP / (TP + FN)`
- **설명:** 실제 Positive 중 모델이 Positive로 예측한 비율
- **범위/해석:** 0~1, 1에 가까울수록 좋음
- **장점:** 
  - False Negative 최소화 중요할 때 필수
  - 실제 양성 사례 놓치지 않음
- **단점:** 
  - 과대예측 편향 가능성
  - Precision과 상충 관계
- **사용예:** 암 진단, 사기 탐지, 보안 시스템

## 4. Specificity (특이도)

- **공식:** `Specificity = TN / (TN + FP)`
- **설명:** 실제 Negative 중 모델이 Negative로 예측한 비율
- **범위/해석:** 0~1, 1에 가까울수록 좋음
- **장점:** 
  - Negative 클래스 식별 능력 측정
  - False Positive 통제 평가
- **단점:** 
  - 불균형 데이터에서 해석 주의
- **사용예:** 정상 샘플 정확한 식별이 중요한 경우

## 5. Negative Predictive Value (NPV, 음성 예측도)

- **공식:** `NPV = TN / (TN + FN)`
- **설명:** 모델이 Negative로 예측한 것 중 실제 Negative인 비율
- **범위/해석:** 0~1, 1에 가까울수록 좋음
- **장점:** 
  - Negative 예측의 신뢰성 측정
  - Precision의 Negative 버전
- **단점:** 
  - 클래스 불균형에 민감
- **사용예:** Negative 예측 정확성이 중요한 경우

## 6. F1-Score

- **공식:** `F1 = 2 × (Precision × Recall) / (Precision + Recall)`
- **설명:** Precision과 Recall의 조화평균
- **범위/해석:** 0~1, 1에 가까울수록 좋음
- **장점:** 
  - Precision과 Recall 균형 고려
  - 불균형 데이터에 효과적
- **단점:** 
  - 특정 지표 우선순위 반영 어려움
- **사용예:** 불균형 데이터셋, 전반적 성능 평가

## 7. Fβ-Score

- **공식:** `Fβ = (1 + β²) × (Precision × Recall) / (β² × Precision + Recall)`
- **설명:** Precision과 Recall 가중 조화평균, β로 가중치 조절
- **범위/해석:** 0~1, 1에 가까울수록 좋음
- **장점:** 
  - β>1이면 Recall 중시, β<1이면 Precision 중시
  - 상황별 맞춤 평가 가능
- **단점:** 
  - β 값 선정의 주관성
- **사용예:** 특정 지표에 가중치를 두고 싶을 때

## 8. Precision-Recall Curve

- **설명:** 임계값 변화에 따른 Precision과 Recall 관계 시각화
- **범위/해석:** 곡선 아래 면적(PR AUC)이 클수록 좋음
- **장점:** 
  - 불균형 데이터에 ROC보다 유용
  - 임계값별 성능 변화 파악
- **단점:** 
  - 해석이 복잡함
- **사용예:** 불균형 데이터의 모델 성능 비교

## 9. Classification Report

- **설명:** 각 클래스별 Precision, Recall, F1-Score를 종합 표시
- **장점:** 
  - 클래스별 상세 성능 파악
  - 다중 클래스 분석에 유용
- **단점:** 
  - 요약 정보 부족
- **사용예:** 다중 클래스 분류의 세부 성능 분석

## 10. Confusion Matrix (혼동행렬)

- **설명:** 예측 클래스와 실제 클래스의 교차표
- **구성:** TP, TN, FP, FN으로 구성
- **장점:** 
  - 모든 분류 결과 한눈에 파악
  - 다른 지표 계산의 기초
- **단점:** 
  - 클래스 수 증가 시 복잡해짐
- **사용예:** 분류 성능의 기본 분석

## 11. Multilabel Confusion Matrix

- **설명:** 다중 레이블 분류에서 각 레이블별 confusion matrix
- **장점:** 
  - 각 레이블별 성능 개별 분석
  - 레이블간 성능 차이 파악
- **단점:** 
  - 레이블 수만큼 행렬 생성으로 복잡
- **사용예:** 다중 레이블 분류 문제의 상세 분석

## 12. ROC Curve

- **설명:** TPR(Recall) vs FPR 관계 곡선
- **범위/해석:** 곡선이 왼쪽 위로 치우칠수록 좋음
- **장점:** 
  - 임계값 무관한 성능 평가
  - 균형 데이터에 효과적
- **단점:** 
  - 불균형 데이터에서 과대평가 가능[4]
- **사용예:** 균형 데이터의 이진 분류 성능 평가

## 13. AUC (Area Under Curve)

- **공식:** ROC 곡선 아래 면적
- **범위/해석:** 0~1, 0.5는 랜덤, 1에 가까울수록 좋음[4]
- **장점:** 
  - 단일 숫자로 모델 성능 요약
  - 임계값 독립적 평가
- **단점:** 
  - 불균형 데이터에서 오해 소지[4]
- **사용예:** 모델 간 성능 비교, 균형 데이터 평가

## 14. Log Loss

- **공식:** `Log Loss = -Σ[yi×log(pi) + (1-yi)×log(1-pi)] / N`
- **설명:** 예측 확률과 실제 값의 로그 손실
- **범위/해석:** 0 이상, 0에 가까울수록 좋음
- **장점:** 
  - 확률 값 직접 반영
  - 잘못된 예측에 큰 패널티[10]
- **단점:** 
  - 해석이 직관적이지 않음
- **사용예:** 확률 예측 정확도가 중요한 경우

## 15. 이익도표 (Lift Chart)

- **설명:** 분류 모형을 사용하지 않았을 때보다 사용한 경우 얼마나 더 이익이 되는지를 시각적으로 나타낸 그래프
- **공식:** `Lift = 각 구간별 반응률 / 기본향상도(Baseline Lift)`
- **Baseline Lift:** `전체 True 개수 / 전체 데이터 수`
- **범위/해석:** 1 이상, 높을수록 좋음 (1은 랜덤 수준)
- **장점:**
  - 모델이 '확신하는 정도' 반영 가능
  - 비즈니스 관점에서 모델 가치 평가
  - 마케팅 타겟팅 효과 측정
  - 무작위 대비 향상도 직관적 표현
- **단점:** 
  - 구간 설정의 임의성
  - 데이터 대표성에 의존
  - 과적합 위험 시 과대평가 가능
- **사용예:** 
  - 마케팅 캠페인 최적화
  - 고객 세분화 및 타겟팅
  - 불균형 데이터 성능 측정
  - 상위 n% 고객 선별 전략 수립

**작성 방법:**
1. 예측 확률을 기준으로 데이터를 내림차순 정렬
2. 기본향상도(Baseline Lift) 계산 
3. 데이터를 동일 크기 구간으로 분할 (보통 10등분)
4. 각 구간별 반응률과 Lift 값 계산
5. X축: 구간(백분위), Y축: Lift 값으로 차트 작성

**해석:**
- 좋은 모델일수록 Lift가 빠른 속도로 감소
- 상위 구간에서 높은 Lift 값 유지
- 최종적으로 100%에서 Lift = 1로 수렴
- 급격한 감소 직전까지가 강하게 확신할 수 있는 범위

**Gain Chart와의 차이점:**
- **Gain Chart**: 누적 양성 사례 비율 표시
- **Lift Chart**: 각 구간별 무작위 대비 향상도 배수 표시
- Lift Chart는 항상 1로 수렴하는 특징


## 16. MCC (Matthews Correlation Coefficient)

- **공식:** `MCC = (TP×TN - FP×FN) / √[(TP+FP)(TP+FN)(TN+FP)(TN+FN)]`
- **범위/해석:** -1~1, 1에 가까울수록 좋음, 0은 랜덤 수준
- **장점:** 
  - 불균형 데이터에 강건
  - 모든 confusion matrix 요소 고려
- **단점:** 
  - 해석이 복잡함
- **사용예:** 불균형 데이터의 종합적 성능 평가

## 17. QWK (Quadratic Weighted Kappa)

- **공식:** 가중치가 적용된 카파 계수로 순서형 분류에 사용
- **범위/해석:** -1~1, 1에 가까울수록 좋음, 0은 랜덤 수준
- **장점:** 
  - 순서형 클래스 간 거리 고려
  - 평가자 간 일치도 측정
- **단점:** 
  - 계산 복잡
  - 순서형 데이터에만 적용
- **사용예:** 등급 분류, 평점 예측, 의료 진단 등급

**주요 특징:**
- **TP**: True Positive, **TN**: True Negative, **FP**: False Positive, **FN**: False Negative
- **불균형 데이터**: Precision, Recall, F1-Score, MCC, PR-AUC 권장
- **균형 데이터**: Accuracy, ROC-AUC 사용 가능
- **확률 중시**: Log Loss 사용
- **비즈니스 가치**: 이익도표, Lift Chart 활용

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.datasets import load_breast_cancer
from sklearn.metrics import classification_report

x, y = load_breast_cancer()['data'], load_breast_cancer()['target']

#전처리
sc = StandardScaler()
train_X, test_X, train_Y, test_Y = train_test_split(x, y, random_state=42)

train_X = sc.fit_transform(train_X)
test_X = sc.transform(test_X)

#로지스틱 회귀
lr = LogisticRegression().fit(train_X, train_Y)
pred1 = lr.predict(test_X)

# svm
svc = SVC().fit(train_X, train_Y)
pred2 = svc.predict(test_X)

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, log_loss, matthews_corrcoef, cohen_kappa_score, confusion_matrix

title = ['accuracy', 'precision', 'recall', 'f1_score', 'log_loss', 'mcc', 'cohen']
f_list = [accuracy_score, precision_score, recall_score, f1_score, log_loss, matthews_corrcoef, cohen_kappa_score]

model_list = [('logistic', pred1), ('svm', pred2)]

rs = pd.DataFrame(columns=['모델명', '성능지표명', '값'])

for m_name, p in model_list:
    cm = pd.DataFrame(confusion_matrix(test_Y, p), index=['실제0', '실제1'], columns=['예측0', '예측1'])
    print(f'[{m_name}]')
    display(cm)
    for t, ff in zip(title, f_list):
        r = ff(test_Y, p)
        rs.loc[len(rs)] = [m_name, t, r]
        
g1 = rs[rs['모델명'] == 'logistic'].reset_index(drop=True)
g2 = rs[rs['모델명'] == 'svm'].reset_index(drop=True)

display(pd.concat([g1, g2], axis=1))

fig, ax = plt.subplots(1, 1, figsize=(8, 6))
sns.lineplot(data=g1, x='성능지표명', y='값', marker='o', ax=ax, label='logistic')
sns.lineplot(data=g2, x='성능지표명', y='값', marker='D', ax=ax, label='svm')
plt.show()
```

모든 평가지표에서 logistic regression 의 성능이 좋다. 

```python
from sklearn.metrics import precision_recall_curve, roc_auc_score, roc_curve, PrecisionRecallDisplay, RocCurveDisplay

proba = lr.predict_proba(test_X)[:, 1] #레이블 1의 예측확률

# 정밀도-재현율의 차이가 최소가 되는 지점 찾기
precision, recall, thresholds = precision_recall_curve(test_Y, proba)
thresholds = list(thresholds) +[1] # 절편 추가
idx = np.argmin(np.abs(precision - recall)) #가장 최소가 되는 지점의 인덱스
best_thresholds = thresholds[idx]

# roc-auc가 최대가 되는 지점 찾기
fpr, tpr, thresholds2 = roc_curve(test_Y, proba)
auc = roc_auc_score(test_Y, proba)

#youden index(youden's j 통계량): recall - (1-specificity)
j = tpr - fpr
idx2 = np.argmax(j)
best_thresholds2 = thresholds2[idx2]

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].plot(thresholds, precision, label='precision')
ax[0].plot(thresholds, recall, label='racall', ls='--')
ax[0].plot(best_thresholds, recall[idx], marker='o', label='best_threshold')
ax[0].set_title(f'precision-recall trade-off(best: {best_thresholds:.3f} / score: {recall[idx]:.3f})')
ax[0].set_xlabel('probability thresholds')
ax[0].set_ylabel('precision&recall')
ax[0].legend()

ax[1].plot(fpr, tpr, label='AUC')
ax[1].plot([0, 1], [0, 1], color='r', ls='--')
ax[1].plot(fpr[idx2], tpr[idx2], marker='o', label='best_threshold')
ax[1].set_title(f"ROC Curve(best: {best_thresholds2:.3f} / score: {auc:.3f})")
ax[1].legend()
ax[1].set_xlabel('FPR')
ax[1].set_ylabel('TPR')

plt.show()
```

precision-recall 그래프에서 최적의 thresholds가 0.487라는 것은 이 값을 기준으로 0, 1 레이블을 분류하였을 시 정밀도-재현율 차이가  
최소화된다는 것이다.  
roc curve도 마찬가지이다. 

```python
# 새로운 분류기준 후 평가결과
from collections import Counter
new_label = np.where(proba >= best_thresholds, 1, 0) #스레드홀드 기준 분류
Counter(new_label)

for name, fc in zip(title, f_list):
    score = fc(test_Y, new_label)
    print(f"{name}: {score}")
```

```python
#left chart: 예측된 확률과 실제 클래스를 확률 내림차순으로 정렬

rank = pd.DataFrame({'pred_prob':proba, 'true_y': test_Y}).sort_values(by='pred_prob', ascending=False).reset_index(drop=True)

# 10개 구간으로 분류
bins = 10
rank['decile'] = np.abs(11- (pd.qcut(rank['pred_prob'], q=bins, labels=False) + 1))  # 1~10 사이 정수로 표시
rank

#baseline lift 계산 및 실구매자수 집계
total = len(test_Y)
cnt = test_Y.sum() # 실제 1 개수
baseline_lift = cnt/total #전체개수 중 실제 1 비율
print(f"baseline lift: {baseline_lift:.3f}")

liftchart = rank.groupby('decile').sum()

#captured response, response, lift 추가
liftchart['captured_r'] = liftchart['true_y']/cnt
liftchart['r'] = liftchart['true_y']/(total/bins)
liftchart['lift'] = liftchart['r']/baseline_lift
print(f"lift chart:")
display(liftchart)

plt.figure(figsize=(8, 5))
sns.barplot(x=liftchart.index, y=liftchart['lift'])
plt.title('lift chart')
plt.xlabel('decile')
plt.ylabel('lift')
plt.show()
```

상위등급에서 더 높은 반응률을 보이는 모델이 좋은 모델이다.  
도표에서 상위등급인 1의 반응률(response)이 가장 높고, 하위등급으로 갈수록 향상도가 점차 줄어들다가 7에서 급격하게 줄어들어서  
좋은 모형이라고 볼 수 있다.

```python
# multi-class 분류모델 평가
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_wine

x = load_wine()['data'][:, :5] #5개 feature만
y = load_wine()['target']
train_X, test_X, train_Y, test_Y = train_test_split(x, y, random_state=0)

#모델적합
svm = SVC(C=1, kernel='rbf', gamma='auto', probability=True).fit(train_X, train_Y) #proba 출력

pred = svm.predict(test_X)
proba = svm.predict_proba(test_X)

m_list = ['accuracy', 'precision', 'recall', 'f1_score']
f_list = [accuracy_score, precision_score, recall_score, f1_score]

for name, fc in zip(m_list, f_list):
    if name != 'accuracy':
        score = fc(test_Y, pred, average='macro') #클래스 균형
    else:
        score = fc(test_Y, pred)
    print(f"{name} : {score:.3f}")

cm = pd.DataFrame(confusion_matrix(test_Y, pred), index=['실제0', '실제1', '실제2'], columns=['예측0', '예측1', '예측2'])
display(cm)
print(classification_report(test_Y, pred))
```

```python
# 성능 시각화
from sklearn.preprocessing import label_binarize
from sklearn.multiclass import OneVsRestClassifier

train_Y2 = label_binarize(train_Y, classes=np.unique(train_Y)) # one-hot
test_Y2 = label_binarize(test_Y, classes=np.unique(test_Y)) #one-hot

ovr_clf = OneVsRestClassifier(svm) #모델
ovr_clf.fit(train_X, train_Y2)
prob3 = ovr_clf.predict_proba(test_X)

fprs, tprs, aucs, best_thresholds, best_idx = [], [], [], [], []

for i in range(len(np.unique(train_Y))):
    fpr, tpr, thresholds = roc_curve(test_Y2[:, i], prob3[:, i])
    idx = np.argmax(tpr - fpr) #최적
    best_threshold = thresholds[idx]
    auc = roc_auc_score(test_Y2[:, i], prob3[:, i]) #각 클래스별
    fprs.append(fpr)
    tprs.append(tpr)
    aucs.append(auc)
    best_thresholds.append(best_threshold)
    best_idx.append(idx)

fig, ax = plt.subplots(1, 3, figsize=(12, 4))

for i, ax in enumerate(ax):
    ax.plot(fprs[i], tprs[i], label='ROC Curve')
    ax.set_title(f'class {i} (AUC: {aucs[i]:.3f})')
    ax.plot(fprs[i][best_idx[i]], tprs[i][best_idx[i]], marker='o', color='r')
plt.show()
```

3 class 분류모델로 각 레이블 데이터 분포가 불균형은 아니나 가장 적은 수의 데이터를 가진 class 2의 예측을 잘 해내지 못하였다.

- 레이블 1: 재현율은 높으나 정밀도(0으로 예측한 데이터 중 실제 0 비율)는 낮음

# 11-3. 군집모델 평가지표

# 군집모델 평가지표 종합 정리

## 1. 실루엣 계수 (Silhouette Coefficient)

- **공식:** `s(i) = (b(i) - a(i)) / max(a(i), b(i))`
- **설명:** 각 데이터 포인트가 속한 군집 내 유사도와 인접한 군집과의 유사도를 비교
- **범위/해석:** -1~1, 1에 가까울수록 좋음, 0은 경계, 음수는 잘못 할당
- **장점:** 
  - 클러스터링 알고리즘에 독립적
  - 직관적 해석 가능
  - 시각화 가능
- **단점:** 
  - 데이터 양이 많을 때 계산 부담
  - 복잡한 형태 군집에서 한계
- **사용예:** 최적 군집 개수 결정, 알고리즘 성능 비교

## 2. Calinski-Harabasz Score

- **공식:** `CH = (군집 간 분산 / (K-1)) / (군집 내 분산 / (N-K))`
- **설명:** 군집 간 분산과 군집 내 분산의 비율, Variance Ratio Criterion
- **범위/해석:** 0 이상, 클수록 좋음
- **장점:** 
  - 계산 속도 빠름 (O(n) 시간복잡도)
  - 군집 간 분리도와 군집 내 밀집도 동시 고려
- **단점:** 
  - 구형 군집에만 효과적
  - 노이즈에 민감
- **사용예:** 최적 군집 개수 선택, 알고리즘 비교

## 3. Davies-Bouldin Score

- **공식:** `DB = (1/K) × Σ max[(σi + σj) / d(ci, cj)]`
- **설명:** 각 군집과 가장 유사한 군집 간의 평균 거리 계산
- **범위/해석:** 0 이상, 작을수록 좋음
- **장점:** 
  - 실루엣 계수보다 계산량 적음
  - 간단한 계산
- **단점:** 
  - 구형 군집 가정
  - 군집 개수에 민감
- **사용예:** 군집 품질 평가, 최적 K 선택

## 4. ARI (Adjusted Rand Index)

- **공식:** `ARI = (RI - E[RI]) / (max(RI) - E[RI])`
- **설명:** 실제 레이블과 군집 결과 간 유사성 측정, 무작위 기댓값으로 보정
- **범위/해석:** -0.5~1, 1은 완전 일치, 0은 무작위 수준
- **장점:** 
  - 무작위 군집화 보정
  - 클러스터 개수 다른 경우에도 공정한 비교
- **단점:** 
  - 실제 레이블 필요
  - 알고리즘 개발용으로 제한적
- **사용예:** 알고리즘 개발, 성능 비교

## 5. NMI (Normalized Mutual Information)

- **공식:** `NMI = MI(P,Q) / sqrt(H(P) × H(Q))`
- **설명:** 정보이론 기반 척도, 정규화된 상호 정보량
- **범위/해석:** 0~1, 1에 가까울수록 좋음
- **장점:** 
  - 정보이론적 해석
  - 클러스터 개수 변화에 안정적
- **단점:** 
  - 실제 레이블 필요
  - 해석이 복잡
- **사용예:** 알고리즘 성능 평가, 연구 개발

## 6. Completeness Score

- **공식:** `Completeness = 1 - H(P|Q) / H(P)`
- **설명:** 동일 실제 클래스의 샘플이 하나의 군집에 할당되는 정도
- **범위/해석:** 0~1, 1에 가까울수록 좋음
- **장점:** 
  - 직관적 해석
  - 클래스 순수도 측정
- **단점:** 
  - 실제 레이블 필요
  - Homogeneity와 상충 관계
- **사용예:** 군집 순수도 평가

## 7. Homogeneity Score

- **공식:** `Homogeneity = 1 - H(Q|P) / H(Q)`
- **설명:** 각 군집이 단일 클래스의 샘플만 포함하는 정도
- **범위/해석:** 0~1, 1에 가까울수록 좋음
- **장점:** 
  - 군집 내 균일성 측정
  - 이해하기 쉬움
- **단점:** 
  - 실제 레이블 필요
  - Completeness와 균형 필요
- **사용예:** 군집 내 일관성 평가

## 8. V-Measure

- **공식:** `V = 2 × (Homogeneity × Completeness) / (Homogeneity + Completeness)`
- **설명:** Homogeneity와 Completeness의 조화평균, F1-Score와 유사
- **범위/해석:** 0~1, 1에 가까울수록 좋음
- **장점:** 
  - 균형잡힌 종합 평가
  - F1-Score와 유사한 해석
- **단점:** 
  - 실제 레이블 필요
  - 두 지표 모두 높아야 함
- **사용예:** 종합적 군집 품질 평가

## 9. Rand Index

- **공식:** `RI = (a + b) / C(n,2)`
- **설명:** 모든 데이터 쌍에 대해 군집 할당 일치도 측정
- **범위/해석:** 0~1, 1에 가까울수록 좋음
- **장점:** 
  - 계산 간단
  - 직관적 이해
- **단점:** 
  - 무작위 군집화에서도 높은 값
  - 보정되지 않은 지표
- **사용예:** ARI 계산의 기초, 기본적 유사도 측정

## 10. Mutual Information

- **공식:** `MI(P,Q) = ΣΣ P(i,j) × log(P(i,j) / (P(i) × P(j)))`
- **설명:** 두 군집 할당 간 공유하는 정보량
- **범위/해석:** 0 이상, 클수록 좋음
- **장점:** 
  - 정보이론적 기반
  - 비선형 관계 포착
- **단점:** 
  - 정규화되지 않음
  - 군집 개수에 의존적
- **사용예:** NMI, AMI 계산의 기초

**주요 특징:**
- **내부 평가 지표**: 실루엣 계수, Calinski-Harabasz, Davies-Bouldin (실제 레이블 불필요)
- **외부 평가 지표**: ARI, NMI, Completeness, Homogeneity, V-Measure, Rand Index, MI (실제 레이블 필요)
- **최적 군집 개수 결정**: 실루엣 계수, Calinski-Harabasz, Davies-Bouldin 주로 사용
- **알고리즘 비교**: 외부 평가 지표들이 더 신뢰성 있음

```python
from sklearn.datasets import load_iris

x, y = load_iris()['data'], load_iris()['target']

from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3).fit(x)
cluster = kmeans.predict(x)
```

```python
# 실제 군집값이 없는 경우: 비지도학습
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score
f_list = [silhouette_score, calinski_harabasz_score, davies_bouldin_score]
title = ['실루엣 점수(1에 가깝게)', '칼리스키 하나바즈 점수(클수록)', '데이비드볼딘 점수(작을수록)']

for f, t in zip(f_list, title):
    score = f(x, cluster)
    print(f"{t}: {score:.3f}")
```

```python
#실제 군집값이 있는 경우: 실제 레이블이 있는 경우
from sklearn.metrics import (adjusted_rand_score, normalized_mutual_info_score, 
                             completeness_score, homogeneity_score, v_measure_score,  mutual_info_score) 
from scipy.special import comb

def rand_index(true, pred):
    classes, class_idx = np.unique(true, return_inverse=True)
    clusters, cluster_idx = np.unique(pred, return_inverse=True)
    
    n = len(true) #데이터 크기
    
    #교차표
    cr = np.zeros((classes.size, clusters.size), dtype=int)
    for i in range(n):
        cr[class_idx[i], cluster_idx[i]] += 1
    #총 가능한 쌍 수
    total_pairs = comb(n, 2)
    
    #같은 군집에 속하는 쌍 개수
    sum_same_cluster = sum(comb(cr[i, j], 2) for i in range(cr.shape[0]) for j in range(cr.shape[1]))
    
    #각 군집별 가능한 쌍의 수
    sum_a = sum(comb(np.sum(cr, axis=1)[i], 2) for i in range(cr.shape[0]))
    sum_b = sum(comb(np.sum(cr, axis=0)[j], 2) for j in range(cr.shape[1]))
        
    RI = (sum_same_cluster + (total_pairs - sum_a - sum_b + sum_same_cluster)) / total_pairs
    
    return RI


f_list = [adjusted_rand_score, normalized_mutual_info_score, completeness_score, homogeneity_score, 
          v_measure_score, mutual_info_score, rand_index]
title = ['ARI(1)', 'NMI(1)', 'completeness(1)', 'homogeneity(1)', 'v-measure(1)', 'MI(클수록)', 'rand_score(1)']

for f, n in zip(f_list, title):
    score = f(y, cluster)
    print(f"{n}: {score:.3f}")
```

# 11-4. 거리지표

## 1. 유클리디안 거리 (Euclidean Distance)

- **공식:** `d(x,y) = √[Σ(xi - yi)²]`
- **설명:** 두 점 사이의 직선 거리, 피타고라스 정리 기반
- **범위/해석:** 0 이상, 0에 가까울수록 유사
- **장점:** 
  - 가장 직관적이고 일반적
  - 계산이 간단
  - 기하학적 해석 용이
- **단점:** 
  - 고차원에서 의미 약화 (차원의 저주)
  - 이상치에 민감
  - 스케일에 영향받음
- **사용예:** KNN, K-means 클러스터링, 일반적인 거리 측정

## 2. 표준화 거리 (Standardized Distance)

- **공식:** `d(x,y) = √[(x-y)ᵀD⁻¹(x-y)]`, D = diag(s₁₁,...,sₚₚ)
- **설명:** 각 변수를 표준편차로 척도 변환 후 유클리디안 거리 계산
- **범위/해석:** 0 이상, 0에 가까울수록 유사
- **장점:** 
  - 척도 차이로 인한 왜곡 방지
  - 변수별 분산 고려
  - 다양한 단위 변수 통합 분석 가능
- **단점:** 
  - 변수 간 상관관계 무시
  - 표준화 과정에서 정보 손실 가능
- **사용예:** 서로 다른 단위의 변수들을 포함한 분석

## 3. 맨하탄 거리 (Manhattan Distance)

- **공식:** `d(x,y) = Σ|xi - yi|`
- **설명:** 격자형 도로망에서의 거리, L1 거리, 택시 거리
- **범위/해석:** 0 이상, 0에 가까울수록 유사
- **장점:** 
  - 이상치에 덜 민감
  - 계산이 간단
  - 고차원에서도 안정적
- **단점:** 
  - 기하학적 직관성 부족
  - 회전에 민감
- **사용예:** 키패드 거리, 도시 블록 거리, 텍스트 마이닝

## 4. 마할라노비스 거리 (Mahalanobis Distance)

- **공식:** `d(x,y) = √[(x-y)ᵀS⁻¹(x-y)]`, S = 공분산행렬
- **설명:** 변수의 분산과 공분산 구조를 모두 고려한 통계적 거리
- **범위/해석:** 0 이상, 0에 가까울수록 유사
- **장점:** 
  - 변수 간 상관관계 고려
  - 통계적으로 의미있는 거리
  - 이상치 탐지에 효과적
- **단점:** 
  - 공분산 행렬 계산 필요
  - 샘플 크기가 작으면 불안정
  - 계산 복잡도 높음
- **사용예:** 이상치 탐지, 판별분석, 품질관리

## 5. 민코우스키 거리 (Minkowski Distance)

- **공식:** `d(x,y) = [Σ|xi - yi|ᵖ]^(1/p)`
- **설명:** 유클리디안(p=2)과 맨하탄(p=1) 거리를 일반화한 거리
- **범위/해석:** 0 이상, 0에 가까울수록 유사
- **장점:** 
  - 다양한 거리 측도 통합
  - p값 조정으로 유연한 적용
  - 일반화된 프레임워크 제공
- **단점:** 
  - p값 선택의 주관성
  - p가 클수록 계산 복잡
- **사용예:** 머신러닝 알고리즘의 일반화된 거리 함수

## 6. 캔버라 거리 (Canberra Distance)

- **공식:** `d(x,y) = Σ|xi - yi| / (|xi| + |yi|)`
- **설명:** 두 벡터의 차이를 합으로 정규화한 거리
- **범위/해석:** 0~1, 0에 가까울수록 유사
- **장점:** 
  - 상대적 차이 측정
  - 스케일에 덜 민감
  - 0에 가까운 값들의 차이 강조
- **단점:** 
  - 0값에서 정의되지 않음
  - 해석이 복잡
- **사용예:** 생물정보학, 스파스 데이터 분석

## 7. 체비셰프 거리 (Chebyshev Distance)

- **공식:** `d(x,y) = max|xi - yi|`
- **설명:** 각 차원에서 최대 차이값, L∞ 거리, 체스판 거리
- **범위/해석:** 0 이상, 0에 가까울수록 유사
- **장점:** 
  - 가장 다른 차원의 차이 강조
  - 계산이 매우 간단
  - 최악의 경우 고려
- **단점:** 
  - 다른 차원 정보 무시
  - 이상치에 매우 민감
- **사용예:** 게임 이론, 체스 킹의 이동 거리, 최적화 문제

**주요 특징:**
- **L1 거리**: 맨하탄 거리
- **L2 거리**: 유클리디안 거리  
- **L∞ 거리**: 체비셰프 거리
- **통계적 거리**: 표준화 거리, 마할라노비스 거리
- **스케일 의존성**: 유클리디안, 맨하탄, 체비셰프 거리는 스케일에 의존
- **고차원 적합성**: 맨하탄 거리가 유클리디안보다 고차원에서 안정적
- **이상치 민감도**: 유클리디안 > 체비셰프 > 맨하탄 순으로 민감

```python
from scipy.spatial.distance import euclidean, seuclidean, cityblock, minkowski, mahalanobis

nv1 = np.array([1, 5, 7 ,9])
nv2 = np.array([2, 4, 8, 13])
v = np.array([0.1, 1, 10, 5]) #분산
lv = np.array([[1, 0.5, 0.5, 1], [0.1, 2, 1, 0.5], [0.5, 0.5, 1, 0.2], [1, 1, 1, 1]]) #표본공분산

eu1 = euclidean(nv1, nv2)
eu2 = np.sqrt(((nv1- nv2)**2).sum())

print(f"유클리디안 거리: {eu1}(={eu2})")
```

```python
seu1 = seuclidean(nv1, nv2, v)
seu2 = np.sqrt(((nv1 - nv2)**2/v).sum())

print(f"표준화 거리: {seu1}(={seu2})")
```

```python
man1 = cityblock(nv1, nv2)
man2 = np.abs(nv1 - nv2).sum()

print(f"맨해튼 거리: {man1}(={man2})")
```

```python
p = 1
min1 = minkowski(nv1, nv2, p) #맨해튼
min2 = np.abs(nv1 - nv2).sum()**1/p

p = 2
min3 = minkowski(nv1, nv2, p) #유클리디안
min4 = np.sqrt(((nv1- nv2)**2).sum())

print(f"민코스키 거리(p=1): {min1}")
print(f"민코스키 거리(p=2): {min3}")
```

```python
mah1 = mahalanobis(nv1, nv2, lv)

mn = len(nv1)
re_mat1 = (nv1-nv2).reshape(-1, mn) # 1, 4
rs1 = np.dot(re_mat1, lv) #행렬곱
re_mat2 = (nv1-nv2).reshape(-mn, 1) # 4,1
rs2 = np.dot(rs1, re_mat2)

mah2 = float(np.sqrt(rs2))

print(f"마할라노비스 거리: {mah1}(={mah2})")
```

### 자카드 거리 (Jaccard Distance)

| 구분     | 내용                                                                                       |
|---------|--------------------------------------------------------------------------------------------|
| **공식**  | $$\displaystyle d(x,y) = 1 - \frac{|X \cap Y|}{|X \cup Y|}$$ <br>(집합 $X, Y$의 경우)            |
| **설명**  | 두 집합 간의 **교집합 크기 대비 합집합 크기 비율**의 보수, <br> 즉 “겹치는 정도”의 역값을 거리로 정의. 0에 가까울수록 유사한 집합 |
| **범위/해석** | 0 이상 1 이하, 0에 가까울수록 유사                                        |
| **장점**  | - 이진(집합) 데이터에 적합<br>- 희소 데이터 비교에 강함<br>- 불필요한 중복 감쇄 효과                  |
| **단점**  | - 수치형 데이터 직접 적용 어려움<br>- 교집합과 합집합 정의 필요<br>- 크기가 작은 집합은 영향 큼              |
| **사용예** | 텍스트 마이닝(문서 유사도), 추천시스템, 생물정보학(유전자 집합 비교)                           |

### 코사인 거리 (Cosine Distance)

| 구분     | 내용                                                                                                                 |
|---------|----------------------------------------------------------------------------------------------------------------------|
| **공식**  | $$\displaystyle d(x,y) = 1 - \frac{x \cdot y}{\|x\| \|y\|} = 1 - \cos\theta$$ (두 벡터 사이의 코사인 유사도의 보수)       |
| **설명**  | 두 벡터가 이루는 각도의 코사인 값을 통한 유사도 → 거리는 유사도의 보수. 방향 중심 비교에 적합하며 크기(길이) 영향 최소화 |
| **범위/해석** | 0 이상 2 이하 (비음수가정 시 0~1 범위 일반적), 0에 가까울수록 방향이 유사                                  |
| **장점**  | - 크기보다 방향성을 반영<br>- 희소·고차원 데이터에 적합<br>- 정보 검색, 문서 분류 등에서 널리 활용                     |
| **단점**  | - 음수 값 있을 경우 해석 복잡<br>- 각도만 고려해 크기 차이 무시<br>- 경우에 따라 실제 거리 의미와 다를 수 있음        |
| **사용예** | 문서 유사도 계산, 텍스트 마이닝, 추천 시스템, 자연어처리, 이미지 분석                                           |


```python
from scipy.spatial.distance import jaccard, cosine
from sklearn.metrics import pairwise_distances

bdf = pd.DataFrame({'Life':[1, 1], 'Life2':[0,1], 'Life3':[0,1], 'Life4':[0,1], 'Love':[0,1], 'Love2':[0,1], 'Love3':[0,1], 'Love4':[0,1], 'Love5':[0,1], 'Love6':[0,1], 'Love7':[0,1], 'Learn':[1,1], 'Learn2':[1,1], 'Learn3':[1,1], 'Learn4':[1,0], 'Learn5':[1,0] }, index=['doc1', 'doc2'])


doc1 = np.array(bdf.loc['doc1'])
doc2 = np.array(bdf.loc['doc2'])

display(bdf)
```

```python
# 자카드 거리 및 유사도 계산
jac1 = jaccard(doc1, doc2)
jac2 = pairwise_distances(bdf.values, metric='jaccard')[0][1]

check = pd.DataFrame(bdf.sum(), columns=['check'])
union = len(check[check['check'] != 0]) #합집합
intersec = len(check[check['check'] == 2]) #교집합
jac3 = 1 - intersec/union

print(f"자카드 거리: {jac1} (={jac2}, {jac3})")
print(f"자카드 유사도: {intersec/union}")
```

```python
#빈도 데이터
base = ['Life', 'Love', 'Learn']

rs = pd.DataFrame(index=['doc1', 'doc2'])
for w in base:
    tt = []
    for col in bdf.columns:
        if w in col:
            tt.append(col)
    tmp = pd.DataFrame(bdf[tt].sum(axis=1).values, columns=[w], index=['doc1', 'doc2'])
    rs = pd.concat([rs, tmp], axis=1)
display(rs)

doc1 = np.array(rs.loc['doc1'])
doc2 = np.array(rs.loc['doc2'])
```

```python
cos1 = cosine(doc1, doc2)
cos2 = pairwise_distances(rs.values, metric='cosine')[0][1]

cossim = np.sum(doc1*doc2) / (np.sqrt(np.sum(doc1**2)) * np.sqrt(np.sum(doc2**2)))
cos3 = 1 - cossim

print(f"코사인 거리: {cos1} (={cos2}, {cos3})")
print(f"코사인 유사도: {cossim}")
```

- MSE: 실제값과 예측값의 차이를 제곱해 평균을 계산(지표가 직관적이나 스케일 의존적, 이상치 민감)
- MAPE: MAE를 퍼센트 변환(직관적이며 이해하기 쉬우나 실제값에 0이 포함될 경우 사용 불가)

```python
#간단하게 선형모델 구현 후 테스트

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error

cat_list = ['area_type', 'availability', 'size', 'bath', 'balcony']

x = real.drop(columns=['price'])
y = real['price']

x = pd.get_dummies(x, columns=cat_list, drop_first=True)

train_X, test_X, train_Y, test_Y = train_test_split(x, y, random_state=42)

lr = LinearRegression()
lr.fit(train_X, train_Y)
pred = lr.predict(test_X)

mae = mean_absolute_error(test_Y, pred)
mse = mean_squared_error(test_Y, pred)
rmse = np.sqrt(mse)
mape = (np.abs((test_Y - pred)/test_Y)).mean()*100

rs = [mae, mse, rmse, mape]
title = ['mae', 'mse', 'rmse', 'mape']

for s, t in zip(rs, title):
    print(f"{t}: {s}")
```

2. 다음과 같은 데이터로 'cust_id'를 적절한 군집으로 할당하는 모델을 생성한다면, 어떤 평가지표로 평가할 수 있는지 2개 이상의  
   평가지표를 예로 들어 설명하시오.

```python
dt = pd.read_csv('https://raw.githubusercontent.com/algoboni/pythoncodebook1-1/main/practice10_credit_card.csv')
display(dt.head(3))
```

레이블이 제시되어 있지 않으므로 다음의 평가지표로 평가
- 실루엣 계수: 군집 내의 응집도와 군집 간 분리도를 이용한 지표로 군집 내 거리가 짤고 군집 간 거리가 멀수록 좋음(완벽한 군집화: 1)
- davies-bouldin score: 가장 유사한 군집이 있는 각 군집의 평균 유사성 측정값으로 유사성은 군집내 거리와 군집간 거리의 비율
- calinski and harabasz score: 클러스터 내 분산과 클러스터 간 분산의 비율(클수록 좋음)

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score

x = dt.drop(columns=['CUST_ID'])
kmeans = KMeans(n_clusters=3)
kmeans.fit(x)

cluster = kmeans.predict(x)

f_list = [silhouette_score, davies_bouldin_score, calinski_harabasz_score]
title = ['실루엣', '데이비즈', '칼린스키']

for f, t in zip(f_list, title):
    score = f(x, cluster)
    print(f"{t}: {score:.3f}")
```

실루엣 계수는 1에 가까울 수록, 칼린스키 하라바즈 점수는 값이 클수록, 데이비즈 볼린 점수는 값이 낮을수록 좋은 성능을 낸다.

```python
plt.figure(figsize=(10, 5))
sns.countplot(cluster)
plt.show()
```

3. 다음과 같은 데이터 문서1과 문서2의 유사성을 판단할 때 어떤 거리지표를 사용해야 하는지 설명하시오.

```python
from pandas import DataFrame
table = DataFrame([[0,1,0,1,0,0,0,0], [1,1,0,1,0,0,1,1]], index=['문서1', '문서2'], columns=['a', 'hello', 'one', 'is', 'temperature', 'you', 'I', 'banana'])
display(table)
```

범주형 변수의 유사성을 측정할 때는 자카드 거리 또는 코사인 거리를 사용한다.   
- 자카드 유사도: 두 개체가 유사할수록 1, 다를수록 0
- 코사인 거리: 서로 같은수록 1, 다를수록 -1

```python
from sklearn.metrics import pairwise_distances

jac = pairwise_distances(table.values, metric='jaccard')[0][1]
cos = pairwise_distances(table.values, metric='cosine')[0][1]

print(f"자카드 거리: {jac}")
print(f"코사인 거리: {cos}")
```
