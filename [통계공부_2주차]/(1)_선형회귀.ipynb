## 문제 1

다음과 같은 OLS 모델을 만들고 결과를 확인합니다.
- mpg 데이터를 사용하여, 상수항을 포함한 OLS 모델을 생성합니다.
- train, test 데이터는 300, 92개를 각각 사용합니다.
- 독립변수는 ["cylinders", "displacement", "horsepower", "weight", "acceleration", "model_year"]를 사용합니다.
- 종속변수는 "mpg"를 사용합니다.

```python
url = "https://raw.githubusercontent.com/Soyoung-Yoon/bigdata/main/mpg.csv"

df = pd.read_csv(url)
df
```

```python
## <수동방식>
```

```python
from statsmodels.api import OLS, add_constant

X = df[["cylinders", "displacement", "horsepower", "weight", "acceleration", "model_year"]]
y = df["mpg"]
```

```python
### 스케일링 
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()
ss_array = ss.fit_transform(X)
X_scaled = pd.DataFrame(ss_array,
                        columns=X.columns)
X_scaled = add_constant(X_scaled)
```

```python
### 데이터 분할
from sklearn.model_selection import train_test_split

train_input, test_input, train_target, test_target = train_test_split(X_scaled, y,
                                                                     test_size=0.2,
                                                                     random_state=42)
print(train_input.shape, test_input.shape, train_target.shape, test_target.shape)
```

```python
### 다중선형회귀 모델 적합
model = OLS(train_target, train_input).fit()

model.summary()
```

```python
### 평가
from sklearn.metrics import mean_absolute_error

pred = model.predict(test_input)

mae = mean_absolute_error(test_target, pred)
mae
```

## 문제 3

다음과 같은 OLS 모델을 만들고 결과를 확인합니다.
- mpg 데이터를 사용하여, 상수항을 포함한 OLS 모델을 생성합니다.
- train, test 데이터는 300, 92개를 각각 사용합니다.
- 독립변수는 ["cylinders", "displacement", "horsepower", "weight", "acceleration", "model_year"]를 사용합니다.
- 종속변수는 "mpg"를 사용합니다.

```python
url = "https://raw.githubusercontent.com/Soyoung-Yoon/bigdata/main/mpg.csv"

df = pd.read_csv(url)
df
```

```python
ss = StandardScaler()
df[["cylinders", "displacement", "horsepower", 
    "weight", "acceleration", "model_year"]]= ss.fit_transform(df[["cylinders", "displacement", "horsepower",
                                                                   "weight", "acceleration", "model_year"]])
df
```

```python
train = df.iloc[:300,:]
test = df.iloc[300:,:]
```

```python
### formula 방식으로 풀기
from statsmodels.api import OLS

formula = "mpg ~ cylinders + displacement + horsepower + weight + acceleration + model_year"

model = OLS.from_formula(formula, data=train).fit()
model.summary()
```

```python
### 예측 및 평가
from sklearn.metrics import r2_score

pred = model.predict(test)
r2 = r2_score(test["mpg"], pred)
mae = mean_absolute_error(test["mpg"], pred)

print(f"R2-Score : {r2}")
print(f"MAE : {mae}")
```

## 문제 4

다음과 같은 OLS 모델을 만들고 결과를 확인합니다.
- mpg 데이터를 사용하여, 상수항을 포함한 OLS 모델을 생성합니다.
- train, test 데이터는 300, 92개를 각각 사용합니다.
- 독립변수는 ["cylinders", "displacement", "horsepower", "weight", "acceleration", "model_year", "origin"]를 사용합니다.
- "origin"은 더미변수를 만들어 사용합니다.
- 종속변수는 "mpg"를 사용합니다.

```python
url = "https://raw.githubusercontent.com/Soyoung-Yoon/bigdata/main/mpg.csv"

df = pd.read_csv(url)
df
```

```python
df["origin"].unique()
```

```python
### 데이터 분할
train = df.iloc[:300, :]
test = df.iloc[300:, :]
```

```python
### 기준나라를 정하고 그 기준 나라에서 해당 나라로 변할때 차이가 유의한지
```

```python
### formula 방식으로 모델 적합
from statsmodels.api import OLS

formula = "mpg ~ cylinders + displacement + horsepower + weight + acceleration + model_year + C(origin)"

model = OLS.from_formula(formula, data=train).fit()

model.summary()
```

```python
### 예측 및 평가
from sklearn.metrics import r2_score

pred = model.predict(test)

r2 = r2_score(test["mpg"], pred)
r2
```
