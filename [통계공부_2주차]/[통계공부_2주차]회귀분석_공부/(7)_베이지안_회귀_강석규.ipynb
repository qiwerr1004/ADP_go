## <베이지안 회귀>

Q. A시의 20대 남성 411명을 임의로 추출하여 키, 몸무게, 허리둘레를 조사하여 기록한 데이터이다.      
이 데이터를 이용하여 20대 남성의 키와 허리둘레가 체중에 영향을 미치는지 알아보고자 한다

```python
url = 'https://raw.githubusercontent.com/TKV-JONGIN/DataSet/refs/heads/main/ADP_26_problem7.csv'
df = pd.read_csv(url)
df
```

```python
X = df[["height", "waistline"]]
y = df["weight"]
```

```python
### 베이지안 회귀 (BayesianRidge 방식)
from sklearn.linear_model import BayesianRidge

br = BayesianRidge(
                   max_iter=1000,          # 최대 반복횟수
                   tol= 1e-3,            # 수렴 허용오차
                   alpha_1=0.005,        # 형상 모수 (shape)
                   alpha_2=0.005,        # 비율(=척도) 모수 (rate)
                   lambda_1=0.005,       # 형상 모수 (shape)
                   lambda_2=0.005        # 비율(=척도) 모수 (rate)
                  )
br.fit(X, y)
coef = br.coef_
intercept = br.intercept_

print(f"베이지안 회귀 회귀계수 : {coef}")
print(f"베이지안 회귀 절편 : {intercept}")
```

```python
### 베이지안 회귀 (ARDRegression 방식)
from sklearn.linear_model import ARDRegression

ard = ARDRegression(
                    max_iter=1000,          # 최대 반복횟수
                    tol= 1e-3,            # 수렴 허용오차
                    alpha_1=0.005,        # 형상 모수 (shape)
                    alpha_2=0.005,        # 비율(=척도) 모수 (rate)
                    lambda_1=0.005,       # 형상 모수 (shape)
                    lambda_2=0.005        # 비율(=척도) 모수 (rate)  
                    )
ard.fit(X, y)
coef = ard.coef_
intercept = ard.intercept_

print(f"베이지안 회귀 회귀계수 : {coef}")
print(f"베이지안 회귀 절편 : {intercept}")
```
