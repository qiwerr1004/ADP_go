# Anderson-Darling Test : K-S Test보다 꼬리(tail)에 민감해서 실제 데이터에 더 강력한 정규성 검정 가능

```python
# 다양한 분포(normal, exponential, logistic 등)에 대해 검정 가능

# H0: 데이터는 특정 분포를 따른다. 
# H1 : 데이터는 해당 분포를 따르지 않는다.
```

```python
# 통계량 > 임계값 → 귀무가설 기각 (정규분포가 아니다)

import numpy as np
from scipy.stats import anderson
import matplotlib.pyplot as plt

# 1. 예제 데이터 (정규 분포 기반)
np.random.seed(42)
data = np.random.normal(loc=0, scale=1, size=200)

# 2. Anderson-Darling Test 수행
result = anderson(data, dist='norm')  # 정규분포 기준 검정, 'norm', 'expon', 'logistic', 'gumbel', 'extreme1' 등 있음

# 3. 출력 결과
print("Anderson-Darling 통계량:", result.statistic)
print("\n유의수준별 임계값 및 판단:")

# 3. 유의수준 5% 임계값 찾기
target_sig_level = 5.0
critical_value = None

for cv, sig in zip(result.critical_values, result.significance_level):
    if sig == target_sig_level:
        critical_value = cv
        break

# 4. 결과 판단
print("분포 가정: 정규분포 (norm)")
print(f" - 통계량: {result.statistic:.4f}")
print(f" - 유의수준 5% 임계값: {critical_value:.4f}")
```

```python
# 귀무 : 데이터는 지수분포(exponential distribution) 를 따른다.
# 통계량 > 임계값 → 귀무가설 기각 (지수분포가 아니다)

import numpy as np
from scipy.stats import anderson

# 1. 지수분포 데이터 생성 (예시)
np.random.seed(42)
data = np.random.exponential(scale=2.0, size=200)

# 2. Anderson-Darling Test for exponential
result = anderson(data, dist='expon')

# 3. 유의수준 5% 임계값 찾기
target_sig_level = 5.0
critical_value = None

for cv, sig in zip(result.critical_values, result.significance_level):
    if sig == target_sig_level:
        critical_value = cv
        break

# 4. 결과 판단
print("분포 가정: 지수분포 (expon)")
print(f" - 통계량: {result.statistic:.4f}")
print(f" - 유의수준 5% 임계값: {critical_value:.4f}")
```
